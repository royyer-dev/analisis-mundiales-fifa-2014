```python
import pandas as pd

# El "../" le indica a Python que suba un nivel en las carpetas antes de buscar "01_Datos"
df_partidos = pd.read_csv("../01_Datos/Crudos/matches.csv")
df_jugadores = pd.read_csv("../01_Datos/Crudos/players.csv")
df_copamundial = pd.read_csv("../01_Datos/Crudos/worldcups.csv")
```

```python
# 2. Revisión rápida de las dimensiones de nuestros DataFrames
print("Filas/columnas partidos: ", df_partidos.shape)
print("Filas/columnas jugadores: ", df_jugadores.shape)
print("Filas/columnas copa mundial: ", df_copamundial.shape) # Corregido: antes leía df_partidos
```

```python
# 3. Limpieza y transformación de la tabla de Jugadores (Lógica intacta)

# Eliminamos duplicados de origen
df_jugadores = df_jugadores.drop_duplicates()

# Rellenamos nulos
df_jugadores['Position'] = df_jugadores['Position'].fillna('Regular')
df_jugadores['Event'] = df_jugadores['Event'].fillna('Sin_eventos')

# Separamos los eventos múltiples y explotamos la lista para crear una fila por evento
df_jugadores['Lista_Eventos'] = df_jugadores['Event'].str.split(r'\s+')
df_jugadores = df_jugadores.explode('Lista_Eventos')

# Extraemos el tipo de evento y el minuto usando tu misma expresión regular
df_jugadores[['Tipo_Evento', 'Minuto_Evento']] = df_jugadores['Lista_Eventos'].str.extract(r'([A-Za-z_]+)\s*(\d+)?')

# Estandarizamos textos y convertimos los minutos a enteros
df_jugadores['Tipo_Evento'] = df_jugadores['Tipo_Evento'].replace({'Sin_eventos': 'Sin eventos'})
df_jugadores['Minuto_Evento'] = df_jugadores['Minuto_Evento'].fillna(0).astype(int)

# Limpiamos las columnas de apoyo
df_jugadores = df_jugadores.drop(columns=['Event', 'Lista_Eventos'])
```

```python
# 4. Limpieza de la tabla Partidos

# Tratamos los nulos en la asistencia y la convertimos a número entero
df_partidos['Attendance'] = df_partidos['Attendance'].fillna(0).astype(int)

# Eliminamos todas las columnas que no aportan a nuestras preguntas de negocio
df_partidos = df_partidos.drop(columns=[
    'Half-time Home Goals',
    'Half-time Away Goals', 
    'Win conditions',
    'Referee',
    'Assistant 1',
    'Assistant 2'
])
```

```python
# 5. Exportamos los datos procesados (con el salto de carpeta ../ incluido)
df_partidos.to_csv('../01_Datos/Limpios/matches_limpio.csv', index=False)
df_jugadores.to_csv('../01_Datos/Limpios/players_limpio.csv', index=False)
df_copamundial.to_csv('../01_Datos/Limpios/worldcups_limpio.csv', index=False)

print("Archivos creados y guardados")
```
